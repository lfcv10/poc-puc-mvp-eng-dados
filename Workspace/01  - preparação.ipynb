## Preparação

Notebook criado com o intuito de criar o catalogo e as 4 camadas utilizadas. Criadas camadas para:

staging: recebimento dos dados em formato de arquivos
bronze: armazenar os dados no estado original no formato de tabelas
silver: Padronizar e enriquecer as informações.
gold: Transforma os dados em visões consolidadas, tabelas de fatos e dimensões (esquema em estrela) otimizadas para performance.
