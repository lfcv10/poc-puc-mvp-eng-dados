
# Camada Bronze

Notebook criado com o intuito de fazer a transformação dos arquivos contidos no schema de staging em tabelas.

Foram inicialmente criadas três tabelas:

1) resultados_saeb: resultados por escola do Saeb 2023
2) dados_socioeconomicos_estados: Dados do IBGE com principais indicadores socioeconômicos dos estados
3) atributos_estados: Base dimensão com o nome dos estados e região do Brasil associada a cada um


## 1. Preparação

Seleção do catalogo e esquema bronze para recebimento das bases

```python
spark.sql("USE CATALOG `mvp-eng-dados-puc-rio`")
spark.sql("USE SCHEMA `bronze`")
```

## 2. Criação da tabela com os resultados do exame Saeb (resultados_saeb)

```python
df_saeb = spark.read.csv("/Volumes/mvp-eng-dados-puc-rio/staging/dados_saeb/TS_ESCOLA.csv", header=True,sep=";",inferSchema=True)
```

```python
display(df_saeb.limit(10))
```

```python
df_saeb.write.format("delta").mode("overwrite").saveAsTable("resultados_saeb")
```

## 3. Criação da tabela de dados socioeconômicos dos estados (dados_socioeconomicos_estados)

```python
df_estados = spark.read.option("headerRows", 1).excel("/Volumes/mvp-eng-dados-puc-rio/staging/dados_estados/censo_total_1991_2010.xlsx")
```

```python
display(df_estados.limit(10))
```

```python
df_estados.write.format("delta").mode("overwrite").saveAsTable("dados_socioeconomicos_estados")
```

## 4. Criação da tabela de dados gerais dos estados (atributos_estados)

Tabela dimensão com informações da região ao qual o estado pertence.

```python
df_atributos_estados = spark.read.option("headerRows", 1).excel("/Volumes/mvp-eng-dados-puc-rio/staging/dados_estados/atributos_estados.xlsx")
```

```python
display(df_atributos_estados.limit(10))
```

```python
df_atributos_estados.write.format("delta").mode("overwrite").saveAsTable("atributos_estados")
```
