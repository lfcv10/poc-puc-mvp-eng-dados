## 1.1. Função de leitura de catalógos

```python
def documentar_tabela(schema, table_name):
    df_documentacao = spark.sql(f"""
        SELECT 
            table_schema,
            table_name,
            column_name,
            data_type,
            comment
        FROM 
            `mvp-eng-dados-puc-rio`.information_schema.columns
        WHERE 
            table_schema = '{schema}' 
            AND table_name = '{table_name}'
    """)
    
    display(df_documentacao)
    
    return df_documentacao
```
